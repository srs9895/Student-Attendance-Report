```python

print("=" * 40)
print("     STUDENT ATTENDANCE CALCULATOR")
print("=" * 40)

# Step 1: Get student details
student_name = input("Enter the student name: ").strip()

try:
    total_classes = int(input("Enter the total classes: "))
    classes_attended = int(input("Enter the classes attended: "))

    # Step 2: Validate the inputs
    if not student_name:
        print("Error: Student name cannot be empty.")

    elif total_classes <= 0:
        print("Error: Total classes must be greater than zero.")

    elif classes_attended < 0 or classes_attended > total_classes:
        print("Error: Attended classes must be between 0 and total classes.")

    else:
        # Step 3: Calculate attendance
        absent_classes = total_classes - classes_attended
        percentage = (classes_attended / total_classes) * 100

        # Step 4: Determine attendance status
        if percentage >= 75:
            status = "Attendance requirement met"
            message = "Great job! Keep it up!"
        else:
            status = "Attendance requirement not met"
            message = "Try to attend more classes."

        # Step 5: Display the report
        print("\n" + "=" * 40)
        print("          ATTENDANCE REPORT")
        print("=" * 40)

        print(f"Student Name    : {student_name}")
        print(f"Total Classes   : {total_classes}")
        print(f"Classes Attended: {classes_attended}")
        print(f"Classes Absent  : {absent_classes}")
        print(f"Attendance      : {percentage:.2f}%")

        # Simple visual progress bar
        filled = round(percentage / 5)
        progress_bar = "█" * filled + "░" * (20 - filled)
        print(f"Progress        : [{progress_bar}]")

        print(f"Status          : {status}")
        print(f"Message         : {message}")
        print("=" * 40)

except ValueError:
    print("Error: Please enter whole numbers for the class counts.")
```

```python

import math
import ipywidgets as widgets
from IPython.display import display

# ---------- HEADER ----------
header = widgets.HTML("""
<div style="
    background: linear-gradient(135deg, #172554, #2563eb);
    padding: 24px;
    border-radius: 14px;
    color: white;
    font-family: Arial;
">
    <h1 style="margin:0;">🎓 Student Attendance</h1>
    <p style="margin:8px 0 0;">
        Attendance Management Dashboard
    </p>
</div>
""")

# ---------- INPUT FIELDS ----------
student_name = widgets.Text(
    description="Student:",
    placeholder="Enter student name",
    layout=widgets.Layout(width="100%"),
    style={"description_width": "110px"}
)

total_classes = widgets.BoundedIntText(
    value=40,
    min=1,
    max=10000,
    description="Total classes:",
    layout=widgets.Layout(width="100%"),
    style={"description_width": "110px"}
)

attended_classes = widgets.BoundedIntText(
    value=32,
    min=0,
    max=10000,
    description="Attended:",
    layout=widgets.Layout(width="100%"),
    style={"description_width": "110px"}
)

target = widgets.IntSlider(
    value=75,
    min=50,
    max=100,
    step=5,
    description="Target:",
    readout_format="d",
    style={"description_width": "110px"},
    layout=widgets.Layout(width="100%")
)

# ---------- BUTTONS ----------
calculate_btn = widgets.Button(
    description="Calculate Attendance",
    icon="calculator",
    button_style="primary",
    layout=widgets.Layout(width="68%", height="42px")
)

reset_btn = widgets.Button(
    description="Reset",
    icon="refresh",
    layout=widgets.Layout(width="28%", height="42px")
)

# ---------- RESULT AREA ----------
result = widgets.HTML()

def calculate(button):
    name = student_name.value.strip()
    total = total_classes.value
    attended = attended_classes.value
    goal = target.value

    if not name:
        result.value = """
        <div style="padding:14px; background:#fee2e2;
        color:#991b1b; border-radius:10px;">
        Please enter the student's name.
        </div>
        """
        return

    if attended > total:
        result.value = """
        <div style="padding:14px; background:#fee2e2;
        color:#991b1b; border-radius:10px;">
        Attended classes cannot exceed total classes.
        </div>
        """
        return

    percentage = attended / total * 100
    absent = total - attended

    # Calculate additional consecutive classes needed
    # to reach the target percentage.
    if percentage >= goal:
        extra = 0
    else:
        extra = math.ceil(
            (goal * total - 100 * attended) / (100 - goal)
        )

    if percentage >= goal:
        status = "TARGET ACHIEVED"
        status_color = "#15803d"
        message = "Great work! Keep it up."
        bar_color = "#16a34a"
    else:
        status = "BELOW TARGET"
        status_color = "#b45309"
        message = f"Attend the next {extra} classes consecutively to reach your target."
        bar_color = "#f59e0b"

    bar_width = min(percentage, 100)

    result.value = f"""
    <div style="
        font-family:Arial;
        margin-top:16px;
        padding:20px;
        border:1px solid #e2e8f0;
        border-radius:14px;
        color:#1e293b;
    ">
        <p style="color:#64748b; margin:0;">ATTENDANCE REPORT</p>
        <h2 style="margin:8px 0;">{name}</h2>

        <div style="
            background:#f1f5f9;
            border-radius:10px;
            padding:16px;
            margin:12px 0;
        ">
            <p style="margin:0; color:#64748b;">
                Attendance Percentage
            </p>
            <h1 style="margin:6px 0; color:{status_color};">
                {percentage:.2f}%
            </h1>
            <div style="
                height:12px;
                background:#e2e8f0;
                border-radius:20px;
                overflow:hidden;
            ">
                <div style="
                    height:12px;
                    width:{bar_width}%;
                    background:{bar_color};
                    border-radius:20px;
                "></div>
            </div>
            <p style="margin:8px 0 0; color:#64748b;">
                Target: {goal}%
            </p>
        </div>

        <div style="display:flex; gap:10px; flex-wrap:wrap;">
            <div style="
                flex:1; min-width:100px; background:#eff6ff;
                padding:14px; border-radius:10px;
            ">
                <p style="margin:0; color:#1d4ed8;">Total Classes</p>
                <h2 style="margin:6px 0;">{total}</h2>
            </div>
            <div style="
                flex:1; min-width:100px; background:#f0fdf4;
                padding:14px; border-radius:10px;
            ">
                <p style="margin:0; color:#166534;">Present</p>
                <h2 style="margin:6px 0;">{attended}</h2>
            </div>
            <div style="
                flex:1; min-width:100px; background:#fff7ed;
                padding:14px; border-radius:10px;
            ">
                <p style="margin:0; color:#9a3412;">Absent</p>
                <h2 style="margin:6px 0;">{absent}</h2>
            </div>
        </div>

        <div style="
            margin-top:14px;
            padding:14px;
            background:#f8fafc;
            border-radius:10px;
        ">
            <b style="color:{status_color};">{status}</b>
            <p style="margin:8px 0 0;">{message}</p>
        </div>
    </div>
    """

def reset_form(button):
    student_name.value = ""
    total_classes.value = 40
    attended_classes.value = 32
    target.value = 75
    result.value = ""

calculate_btn.on_click(calculate)
reset_btn.on_click(reset_form)

# ---------- DISPLAY DASHBOARD ----------
form = widgets.VBox(
    [
        widgets.HTML("<h3>Enter Student Details</h3>"),
        student_name,
        total_classes,
        attended_classes,
        target,
        widgets.HBox(
            [calculate_btn, reset_btn],
            layout=widgets.Layout(
                justify_content="space-between"
            )
        )
    ],
    layout=widgets.Layout(
        padding="20px",
        border="1px solid #e2e8f0",
        border_radius="14px",
        gap="12px"
    )
)

dashboard = widgets.VBox(
    [header, form, result],
    layout=widgets.Layout(
        width="100%",
        max_width="760px",
        gap="16px"
    )
)

display(dashboard)
```
